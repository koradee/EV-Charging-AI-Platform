## 3 · Datetime Parsing & Duration Verification

```python
# Parse datetime columns
df['start_dt'] = pd.to_datetime(df['Charging Start Time'], errors='coerce')
df['end_dt']   = pd.to_datetime(df['Charging End Time'],   errors='coerce')

# NaT check
n_bad_start = df['start_dt'].isna().sum()
n_bad_end   = df['end_dt'].isna().sum()
print(f'Unparseable Start Times : {n_bad_start}')
print(f'Unparseable End Times   : {n_bad_end}')

# Sample of parsed values
print('\nSample parsed datetimes:')
display(df[['Charging Start Time', 'start_dt', 'Charging End Time', 'end_dt']].head(5))
```

```python
# Compute duration from timestamps (in hours)
df['duration_computed_hrs'] = (df['end_dt'] - df['start_dt']).dt.total_seconds() / 3600

# Difference between computed and recorded duration
df['duration_diff_hrs'] = (df['duration_computed_hrs'] - df['Charging Duration (hours)']).abs()

# Summary of differences
print('Duration difference statistics (|computed - recorded|) in hours:')
display(df['duration_diff_hrs'].describe())

# Flag mismatches: difference > 0.05 hours (~3 minutes tolerance)
TOLERANCE_HRS = 0.05
mismatch_mask = df['duration_diff_hrs'] > TOLERANCE_HRS
n_mismatches = mismatch_mask.sum()
print(f'\nRows with |computed - recorded| > {TOLERANCE_HRS} hrs : {n_mismatches} ({n_mismatches/len(df)*100:.2f}%)')
```

```python
if n_mismatches > 0:
    print('\n️  Mismatch samples (first 20):')
    cols_to_show = [
        'User ID', 'Charging Start Time', 'Charging End Time',
        'Charging Duration (hours)', 'duration_computed_hrs', 'duration_diff_hrs'
    ]
    display(df.loc[mismatch_mask, cols_to_show].head(20))
else:
    print('  All durations match within tolerance — no mismatch flags raised.')
```

```python
# Drop helper columns (keep raw data clean in-memory)
df.drop(columns=['start_dt', 'end_dt', 'duration_computed_hrs', 'duration_diff_hrs'], inplace=True)
print('Helper columns dropped — dataframe back to original 20 columns.')
```

## 5 · Anomaly Detection Pass

```python
# ── 5a. Negative values in numeric columns ──────────────────────────────────────
print('--- Negative Values in Numeric Columns ---')
neg_check = {}
for col in df.select_dtypes(include=[np.number]).columns:
    n_neg = (df[col] < 0).sum()
    if n_neg > 0:
        neg_check[col] = n_neg

if neg_check:
    for col, cnt in neg_check.items():
        print(f'  ️  {col}: {cnt} negative values')
else:
    print('    No negative values found in any numeric column.')
```

```python
# ── 5b. State of Charge out-of-range (0–100%) ──────────────────────────────────
print('--- State of Charge Range Checks ---')
soc_cols = ['State of Charge (Start %)', 'State of Charge (End %)']
for col in soc_cols:
    out = df[(df[col] < 0) | (df[col] > 100)]
    print(f'  {col}: {len(out)} values outside [0, 100]')
    if len(out) > 0:
        display(out[[col]].describe())
```

```python
# ── 5c. SoC End < SoC Start (charging went backwards?) ─────────────────────────
print('--- Sessions where SoC (End) < SoC (Start) ---')
soc_reversed = df[df['State of Charge (End %)'] < df['State of Charge (Start %)']]
print(f'  Count: {len(soc_reversed)}')
if len(soc_reversed) > 0:
    display(soc_reversed[soc_cols + ['Charging Duration (hours)', 'Energy Consumed (kWh)']].head(10))
```

```python
# ── 5d. Zero or near-zero energy/cost/duration ─────────────────────────────────
print('--- Zero or Near-Zero Values ---')
near_zero_cols = [
    'Energy Consumed (kWh)',
    'Charging Duration (hours)',
    'Charging Cost (USD)'
]
for col in near_zero_cols:
    n = (df[col] <= 0).sum()
    print(f'  {col}: {n} zero/negative rows')
```

```python
# ── 5e. Charging Rate vs Energy/Duration consistency ───────────────────────────
print('--- Charging Rate Consistency Check ---')
# Expected: Energy ≈ Rate × Duration  (with some loss tolerance ±20%)
df['rate_check'] = df['Charging Rate (kW)'] * df['Charging Duration (hours)']
df['rate_diff_pct'] = ((df['rate_check'] - df['Energy Consumed (kWh)']).abs() / df['Energy Consumed (kWh)'] * 100)
bad_rate = df[df['rate_diff_pct'] > 20]
print(f'  Sessions where |Rate×Duration - Energy| > 20% of Energy: {len(bad_rate)}')
if len(bad_rate) > 0:
    print('  Sample mismatches:')
    display(bad_rate[[
        'Charging Rate (kW)', 'Charging Duration (hours)',
        'rate_check', 'Energy Consumed (kWh)', 'rate_diff_pct'
    ]].head(10))
df.drop(columns=['rate_check', 'rate_diff_pct'], inplace=True)
```

```python
# ── 5f. Duplicate rows ─────────────────────────────────────────────────────────
print('--- Duplicate Rows ---')
n_dups = df.duplicated().sum()
print(f'  Exact duplicate rows: {n_dups}')

# Same user + same start time (logical duplicate)
n_logical_dups = df.duplicated(subset=['User ID', 'Charging Start Time']).sum()
print(f'  Logical duplicates (same User + Start Time): {n_logical_dups}')
```

```python
# ── 5g. Temperature outliers (plausible range: -30°C to +50°C) ─────────────────
print('--- Temperature Range Check ---')
temp_col = 'Temperature (°C)'
# Note: column name may include encoding artefact; find it
temp_col_actual = [c for c in df.columns if 'emp' in c.lower() and '°' in c or 'Temp' in c][0]
print(f'  Column found: "{temp_col_actual}"')
temp_out = df[(df[temp_col_actual] < -30) | (df[temp_col_actual] > 50)]
print(f'  Values outside [-30°C, 50°C]: {len(temp_out)}')
display(df[temp_col_actual].describe())
```

### Observed Anomalies & Notes

1. **Column encoding artefact** — The `Temperature` column header contains a non-UTF-8 character (`A°C` instead of `°C`). This must be renamed during the cleaning phase.

2. **Duration vs. timestamps** — The recorded `Charging Duration (hours)` appears to be a float with higher precision than what integer-minute timestamps can represent. Expect small mismatches (~seconds level) that are benign; large mismatches (>3 min) should be flagged.

3. **SoC plausibility** — State-of-charge values for both start and end must remain within [0%, 100%]. Any sessions where End SoC < Start SoC without a corresponding negative energy value represent data inconsistencies to be handled.

4. **Charging Rate consistency** — `Charging Rate (kW) × Duration (hours)` should ≈ `Energy Consumed (kWh)`. Significant deviations suggest rate averaging artefacts or session interruptions.

5. **No negative energy/cost/duration values** are expected; confirmed by the anomaly checks above.

6. **Time of Day / Day of Week** are derived categorical labels — they do not add information beyond what timestamps already capture, but are useful for modelling and grouping.
